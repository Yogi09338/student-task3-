```python
# Employee Salary Analysis

salaries = [25000, 32000, 28000, 45000, 38000, 52000, 30000, 41000]

print("Employee Salaries:", salaries)
# Calculate total salary

total_salary = sum(salaries)

print("Total Salary:", total_salary)
# Calculate average salary

average_salary = total_salary / len(salaries)

print("Average Salary:", average_salary)
# Find highest salary

highest_salary = max(salaries)

print("Highest Salary:", highest_salary)
# Find lowest salary

lowest_salary = min(salaries)

print("Lowest Salary:", lowest_salary)
# Salary Analysis Summary

print("Employee Salary Analysis")
print("------------------------")
print("Total Salary:", total_salary)
print("Average Salary:", average_salary)
print("Highest Salary:", highest_salary)
print("Lowest Salary:", lowest_salary)Employee_Salary_Analysis_Yogesh.ipynb
```
